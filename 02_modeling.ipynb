```python
import sys
import os

from sklearn.pipeline import Pipeline

# Add the project root (where the 'src' folder lives) to the path
# If your notebook is in a 'notebooks/' folder, use '../' to go up one level
sys.path.append(os.path.abspath('..')) 
import sys
import os
from pathlib import Path

# Locate the project root containing the src folder
root = Path.cwd()
if not (root / "src").exists():
    for parent in root.parents:
        if (parent / "src").exists():
            root = parent
            break
sys.path.append(str(root))


from src.data_preprocessing import load_and_preprocess
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import classification_report, roc_auc_score, precision_recall_curve
import shap

X_train, X_test, y_train, y_test, preprocessor = load_and_preprocess()

# Full pipeline
xgb_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', XGBClassifier(
        n_estimators=300,
        max_depth=6,
        learning_rate=0.1,
        subsample=0.9,
        colsample_bytree=0.9,
        scale_pos_weight=29,   # Important for imbalance (≈ 9660/339)
        random_state=42,
        eval_metric='aucpr'
    ))
])

xgb_pipeline.fit(X_train, y_train)

# Predictions
y_pred = xgb_pipeline.predict(X_test)
y_prob = xgb_pipeline.predict_proba(X_test)[:, 1]

print(classification_report(y_test, y_pred))
print(f"ROC AUC: {roc_auc_score(y_test, y_prob):.4f}")
```
